```python
import geopandas as gpd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from shapely.geometry import Point

# Load China outline from Natural Earth download
NE_world_countries_path = r"C:\Users\Francesco\Python\earthlab\China\data\ne_110m_admin_0_countries\ne_110m_admin_0_countries.shp"
countries = gpd.read_file(NE_world_countries_path).to_crs(epsg=4326)
china = countries[countries['ADMIN'] == 'China']

# Load Hexigten Banner shapefile
chifeng_path = r"C:\Users\Francesco\OneDrive - University of Leeds\Processing\QGIS\HexigtenBanner\HexigtenBanner.shp"
chifeng = gpd.read_file(chifeng_path).to_crs(epsg=4326)

# Study site point
study_lon = 117.2111
study_lat = 43.3298
study_point = gpd.GeoDataFrame(geometry=[Point(study_lon, study_lat)], crs='EPSG:4326')

# Points to represent the Chinese islands
diaoyu = Point(123.475, 25.748)
xisha = Point(112.33, 16.83)
zhongsha = Point(113.5, 15.8)
nansha = Point(114.0, 10.0)
island_points = gpd.GeoDataFrame(geometry=[diaoyu, xisha, zhongsha, nansha],
                                 crs='EPSG:4326',
                                 index=['Diaoyudao', 'Xisha', 'Zhongsha', 'Nansha'])

# Plot main map
fig = plt.figure(figsize=(10, 10))

main_ax = plt.axes(projection=ccrs.LamberConformal(central_longitude=105, central_latitude=35))
main_ax.set_extent([73, 136, 18, 55], crs=ccrs.PlateCarree())

# Coastline + borders
main_ax.add_feature(cfeature.COASTLINE, linewidth=0.8)
main_ax.add_feature(cfeature.BORDERS, linewidth=0.5)

# China outline
china.boundary.plot(ax=main_ax,
                    edgecolor='black',
                    linewidth=1,
                    zorder=2)

# Chifeng City outline
chifeng.plot(ax=main_ax,
             edgecolor='black',
             facecolor='lightgray',
             linewidth=1.2,
             zorder=3)

# Plot the study area with a star
study_point.plot(ax=main_ax,
                 color='black',
                 marker='*',
                 markersize=50,
                 zorder=4)

# Inset map for South China Sea
inset_ax = fig.add_axes([0.7, 0.08, 0.2, 0.2], projection=ccrs.PlateCarree())
inset_ax.set_extent([105, 125, 3, 25], crs=ccrs.PlateCarree())
inset_ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
inset_ax.add_feature(cfeature.BORDERS, linewidth=0.4)

island_points.plot(ax=inset_ax, )
```
